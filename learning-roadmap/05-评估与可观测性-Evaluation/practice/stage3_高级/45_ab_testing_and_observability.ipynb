```python
MODE = 'OFFLINE'

import numpy as np, hashlib, re, json, time, requests, pandas as pd
from dataclasses import dataclass, field
from typing import Callable, Literal
from datetime import datetime

OLLAMA = 'http://127.0.0.1:11434'

# 模拟 Langfuse 客户端（生产用 langfuse SDK）
class LangfuseStub:
    def __init__(self, mode: Literal['OFFLINE', 'ONLINE'] = 'OFFLINE'):
        self.mode = mode
        self.collected = []
    def observe(self, trace_id: str, input: str, output: str, metadata: dict = None):
        item = {'trace_id': trace_id, 'timestamp': datetime.now().isoformat(),
                'input': input, 'output': output, 'metadata': metadata or {}}
        self.collected.append(item)
        if self.mode == 'ONLINE':
            print(f'  [Langfuse stub] trace={trace_id[:12]}, input={input[:30]}, output={output[:30]}')
    def score(self, trace_id: str, metric_name: str, value: float, comment: str = None):
        item = {'trace_id': trace_id, 'metric_name': metric_name, 'value': value, 'comment': comment}
        self.collected.append(item)
        if self.mode == 'ONLINE':
            print(f'  [Langfuse stub] trace={trace_id[:12]}, metric={metric_name}, value={value}')
    def upload_batch(self):
        if self.mode == 'OFFLINE':
            print(f'  [Langfuse stub] uploaded {len(self.collected)} observations')

LANGFUSE = LangfuseStub(MODE)

def fake_embed(text, dim=256):
    seed = int(hashlib.sha256(text.encode('utf-8')).hexdigest()[:8], 16)
    v = np.random.default_rng(seed).standard_normal(dim).astype(np.float32)
    return v / (np.linalg.norm(v) + 1e-12)

def ollama_embed(text):
    try:
        r = requests.post(f'{OLLAMA}/api/embeddings', json={'model':'nomic-embed-text','prompt':text}, timeout=30)
        r.raise_for_status()
        v = np.array(r.json()['embedding'], dtype=np.float32)
        return v / (np.linalg.norm(v) + 1e-12)
    except Exception as e:
        print(f'[Offline mode] embed failed: {e}')
        return fake_embed(text)

def ollama_chat(prompt, temp=0.0):
    try:
        r = requests.post(f'{OLLAMA}/api/chat', json={
            'model':'qwen1.5_1.8','stream':False,'options':{'temperature':temp},
            'messages':[{'role':'user','content':prompt}]
        }, timeout=120)
        r.raise_for_status()
        return r.json()['message']['content']
    except Exception as e:
        return f'[STUB-LLM] error: {e}'

if MODE == 'ONLINE':
    try:
        requests.get(f'{OLLAMA}/api/tags', timeout=1).raise_for_status()
        embed = ollama_embed; chat = ollama_chat
        print('✅ ONLINE: using real LLM embeddings & chat')
    except Exception:
        MODE='OFFLINE'; embed = fake_embed
        def chat(p): return f'[STUB-LLM] {p[:60]}'
        print('OFFLINE: using fake embed + stub LLM (metrics are rule-based)')

if MODE == 'OFFLINE':
    LANGFUSE.mode = 'OFFLINE'
    print('Langfuse stub: running in OFFLINE mode (no external API calls)')
```

## 1. A/B 测试原理 —— 双轨并行，数据说话

**核心思想**：不同策略并行运行（通常 50/50 流量分配），收集相同用户的**真实反馈**（点击、停留、人工打分），然后用统计方法判断哪个更好。

### A/B 测试 vs Offline Eval
- **Offline eval set**：固定题目，自动化评分（rule-based 或 LLM-as-judge）
- **A/B 测试**：真实用户行为，时间序列数据

### 关键要素
1. **随机分流**：用户随机分配到 Group A 或 Group B（或更复杂的分层分流）
2. **相同评估指标**：必须能横向对比（延迟、准确性、满意度）
3. **统计显著性**：不能说「直觉上好」，要用 p-value /置信区间

### 简单示例
- **实验**：优化 RAG reranker
- **Group A**：dense top-3（baseline）
- **Group B**：dense top-3 + bge-reranker
- **指标**：用户点击率、平均对话轮次、人工满意度评分

```python
# 模拟 A/B 测试流量分流（简单 random split）
def ab_split(user_id: str, group_a: Callable, group_b: Callable) -> dict:
    """为用户分配到 A 或 B 组，返回结果和分配"""
    # 使用 user_id 的 hash 固定分配（同一用户永远在同一组）
    group = 'A' if int(hashlib.md5(user_id.encode()).hexdigest(), 16) % 2 == 0 else 'B'
    
    if group == 'A':
        result = group_a()
        result['variant'] = 'Group A: dense top-3'
    else:
        result = group_b()
        result['variant'] = 'Group B: dense + bge-rerank'
    result['user_id'] = user_id
    result['assigned_group'] = group
    return result

# 模拟两个 RAG 策略（简化版）
DOCS = [
    'XZ4054H 是一款高能量密度电池，容量 300mAh，循环寿命 500 次。',
    'XZ5352R 是一款 800mAh 锂电池，专为高放电场景设计。',
    'XZ4054H-NE1.11 是 XZ4054H 的升级版，集成保护电路，规格书 V1.11。',
    '电源管理芯片 PM8916 提供多路稳压输出，工作电压 3.3V。',
    'Transformer 是 2017 年由 Google 提出的注意力机制神经网络。',
    'RAG 是检索增强生成的简称，把外部知识接进 LLM 上下文。',
    '注意力机制解决了 RNN 在长序列上的梯度衰减问题。',
    'PM8917 是 PM8916 的低功耗替代型号，待机电流降低 50%。',
    'LoRA 是参数高效微调方法，仅训练低秩矩阵 A 和 B。',
    'QLoRA 在 LoRA 基础上把 base 模型 4-bit 量化，单卡可调 7B。',
    'DPO 是无需 reward model 的偏好优化算法。',
    'Chroma 是本地友好的开源向量库，默认 HNSW 索引。',
    'Qdrant 用 Rust 实现，性能优秀，支持稀疏-稠密混合检索。',
    'HNSW 把检索复杂度从 O(N) 降到 O(log N)。',
    'BPE 是字节对编码，被 GPT-2/3/4 等所有现代 LLM 用作分词器。',
]

# 简单 stub reranker
def stub_reranker(query, docs):
    q_tokens = set(re.findall(r'[^\u4e00-\u9fff ]+', query.lower()))
    out = []
    for i, d in enumerate(docs):
        d_tokens = set(re.findall(r'[^\u4e00-\u9fff ]+', d.lower()))
        overlap = len(q_tokens & d_tokens) / (len(q_tokens) + 1e-12)
        out.append((i, overlap))
    return sorted(out, key=lambda x: -x[1])

# Group A: dense only
def group_a():
    q = 'XZ4054H 的升级版是什么'
    vec = embed(q)
    vecs = np.vstack([embed(d) for d in DOCS])
    sims = vecs @ vec
    top_ids = [int(i) for i in np.argsort(-sims)[:3]]
    ctx = [DOCS[i] for i in top_ids]
    answer = chat(f'仅基于上下文回答：\n上下文:\n{'\n'.join(ctx)}\n问题: {q}')
    return {'question': q, 'contexts': ctx, 'answer': answer[:100]}

# Group B: dense + rerank
def group_b():
    q = 'XZ4054H 的升级版是什么'
    vec = embed(q)
    vecs = np.vstack([embed(d) for d in DOCS])
    sims = vecs @ vec
    recall_k = 10
    candidate_ids = [int(i) for i in np.argsort(-sims)[:recall_k]]
    # rerank
    scored = stub_reranker(q, [DOCS[i] for i in candidate_ids])
    top_local = scored[:3]
    final_ids = [candidate_ids[i] for i, _ in top_local]
    ctx = [DOCS[i] for i in final_ids]
    answer = chat(f'仅基于上下文回答：\n上下文:\n{'\n'.join(ctx)}\n问题: {q}')
    return {'question': q, 'contexts': ctx, 'answer': answer[:100]}
```

## 2. Langfuse 集成 —— 评估数据的统一观测平台

**Langfuse** 是开源的「可观测性平台」，专门给 LLM/RAG 工程师设计。
- **Traces**：一次用户请求的完整路径（query → retrieve → generate → feedback）
- **Observations**：每一步的输入/输出
- **Scores**：人工打分或自动化指标

### 集成要点
1. **每条 query 生成一个 trace_id**（可用 user_id + timestamp）
2. **embed/retrieve/generate 都记录 observation**（input → output）
3. **人工反馈打分**（1-5 星）→ 记录 score

### 示例：Langfuse stub 结构

```python
# 模拟用户请求流程
def user_request(user_id: str, question: str, variant_func: Callable):
    trace_id = f'{user_id[:8]}-{datetime.now().strftime("%H%M%S")}'
    
    # 1. 记录 query（trace + observation）
    LANGFUSE.observe(trace_id, input=question, output='...', metadata={'step': 'query'})
    
    # 2. retrieve（模拟）
    vec = embed(question)
    vecs = np.vstack([embed(d) for d in DOCS])
    sims = vecs @ vec
    top_k = 3
    retrieved_ids = [int(i) for i in np.argsort(-sims)[:top_k]]
    contexts = [DOCS[i] for i in retrieved_ids]
    LANGFUSE.observe(trace_id, input=str(retrieved_ids), output=str(contexts),
                     metadata={'step': 'retrieve', 'top_k': top_k})
    
    # 3. generate
    prompt = f'仅基于上下文回答：\n上下文:\n{'\n'.join(contexts)}\n问题: {question}'
    answer = chat(prompt)
    LANGFUSE.observe(trace_id, input=prompt, output=answer, metadata={'step': 'generate'})
    
    # 4. 返回结果（人工反馈字段先设为 None）
    result = {
        'trace_id': trace_id,
        'question': question,
        'contexts': contexts,
        'answer': answer,
        'user_feedback': None,  # 人工打分（1-5）
    }
    return result

# 模拟人工打分（从 Langfuse 外部输入）
def record_human_feedback(trace_id: str, score: int):
    LANGFUSE.score(trace_id, metric_name='user_satisfaction', value=float(score),
                   comment=f'人工评分: {score} 星')

# 模拟 A/B 测试运行
results = []
for i in range(10):
    # 随机分配到 A 或 B 组
    res = ab_split(f'user_{i}', group_a, group_b)
    results.append(res)

print(f'运行了 {len(results)} 条请求，生成 trace_id：')
for r in results:
    print(f'  {r["trace_id"]} -> {r["assigned_group"]}')
```

## 3. 人工评测流程 —— 从「自动化」走向「真实人类」

即使有 LLM-as-judge，**人工评测永远是 RAG 评估的最后一环**。因为：
- LLM-as-judge 模仿的是「人类偏好」，但不是人类本身
- 真实业务场景的边缘情况、细微差别只有人类能判断

### 人工评测流程
1. **准备评测任务**（write eval cases）：覆盖常见问题 + 边缘场景
2. **分配给不同人类**（避免 bias）：每人打 50-100 题
3. **统一标准**（rubric）：评分维度（准确性、完整性、语气等）
4. **合并打分**：求平均，去掉异常值
5. **复盘分析**：看哪些题目大家打分差异大 → 优化 eval set 或模型

### 人工评测 vs 自动化
- **自动化**：快速、批量、低成本
- **人工**：慢、小规模、高成本
- **组合**：自动化筛一遍，人工复查异常和关键 case

```python
# 模拟人工评测流程

# 评测任务集（10 题）
eval_tasks = [
    {'id': 1, 'q': 'XZ4054H 的升级版是什么？', 'gold': 'XZ4054H-NE1.11'},
    {'id': 2, 'q': 'LoRA 训练什么？', 'gold': '低秩矩阵 A 和 B'},
    {'id': 3, 'q': 'PM8916 工作电压是多少？', 'gold': '3.3V'},
    {'id': 4, 'q': 'HNSW 算法复杂度？', 'gold': 'O(log N)'},
    {'id': 5, 'q': '今天纽约天气如何？', 'gold': '不知道'},
    {'id': 6, 'q': 'Transformer 哪年提出的？', 'gold': '2017'},
    {'id': 7, 'q': 'Chroma 是什么？', 'gold': 'HNSW 索引的向量库'},
    {'id': 8, 'q': 'Qdrant 用什么语言写？', 'gold': 'Rust'},
    {'id': 9, 'q': 'bge-reranker 和 dense 有什么区别？', 'gold': 'reranker 精排召回的 docs'},
    {'id': 10, 'q': '平均准确率计算公式？', 'gold': 'correct / total'},
]

# 模拟 3 个人类评测员
evaluators = ['Alice', 'Bob', 'Charlie']
evaluation_results = []

# 1. 评测员 1 打分
print('Alice 正在打分...')
for task in eval_tasks:
    # 这里应该是真实的 UI 让人打分，这里用 stub
    score = 5 if task['gold'] in '好的' else 3  # 简化
    evaluation_results.append({
        'task_id': task['id'],
        'question': task['q'],
        'gold': task['gold'],
        'evaluator': 'Alice',
        'score': score
    })

# 2. 评测员 2 打分
print('Bob 正在打分...')
for task in eval_tasks:
    score = 5 if '好的' in task['gold'] else 3
    evaluation_results.append({
        'task_id': task['id'],
        'question': task['q'],
        'gold': task['gold'],
        'evaluator': 'Bob',
        'score': score
    })

# 3. 评测员 3 打分
print('Charlie 正在打分...')
for task in eval_tasks:
    score = 5 if '好的' in task['gold'] else 3
    evaluation_results.append({
        'task_id': task['id'],
        'question': task['q'],
        'gold': task['gold'],
        'evaluator': 'Charlie',
        'score': score
    })

# 转成 DataFrame
df = pd.DataFrame(evaluation_results)
print(f'\n总评测结果：{len(df)} 条')
print(df.groupby(['task_id', 'evaluator'])['score'].mean())
```

## 4. 可观测性平台总结 —— Langfuse / Datadog / Sentry

### Langfuse 特点
- **开源**：本地部署，可控数据隐私
- **专精 LLM**：trace/observation/score 一体化
- **低成本**：中小规模够用，大厂可自建

### Datadog / Sentry
- **通用 APM 平台**：除了 LLM 也能监控后端服务
- **集成多语言**：Python/Go/Java 都有 SDK
- **可视化**：拓扑图、延迟曲线

### 可观测性 vs 可观测性
两者区别：
- **可观测性**：追踪「它是如何工作的」（log/metric/tracing）
- **可观测性**：追踪「它为何失败」（error tracking + alerting）

### 生产 RAG 的可观测性清单
1. **日志**：每条 query 的完整 trace（Langfuse observation）
2. **指标**：QPS、P95 延迟、用户满意度评分分布
3. **告警**：延迟 > 5s → 告警；满意度 < 3 星 → 告警
4. **回滚**：检测到质量下降 → 自动回滚到上一个稳定版本

## 5. 完整 A/B 测试流程 + 实战建议

### 步骤 1：定义实验目标
- **明确改了什么**：新 reranker、新 prompt、新 chunking 策略
- **定义成功标准**：提升点击率 5%，延迟 < 2s

### 步骤 2：设计实验配置
- **流量分配**：50/50（或更精细的分层）
- **测试时长**：至少 3 天（覆盖工作日/周末）
- **样本量**：至少 1000 用户，5000 次交互

### 步骤 3：埋点采集
- Langfuse 记录每条 trace
- 业务指标：点击、转化、停留时间
- 人工评分：关键 case 抽样打分

### 步骤 4：数据分析
- 对比两组的延迟、准确率、满意度
- 统计显著性检验（t-test）
- 失败 case 分析（看 trace_id 找到具体问题）

### 步骤 5：上线 / 回滚
- 达到目标 → 全量切换
- 质量下降 → 回滚到 A 组

### 实战建议
1. **不要一次改太多**：保持 control group 简单，signal 明显
2. **预发布跑 1 天**：验证流量分流正确
3. **监控 realtime**：每小时看延迟和满意度，防止异常
4. **人工复查关键 case**：LLM-as-judge 可能犯错
5. **记录实验结果**：写 experiment_log.md，为下次实验提供数据

```python
# 模拟 A/B 测试数据收集（1000 用户）
ab_results = []

for i in range(1000):
    user_id = f'user_{i}'
    
    # 分配到 A 或 B 组
    group = 'A' if i % 2 == 0 else 'B'
    
    # 获取结果
    if group == 'A':
        result = group_a()
        latency = 120  # ms
        click = np.random.random() > 0.4  # 60% click rate
    else:
        result = group_b()
        latency = 150  # ms (稍慢，因为有 rerank)
        click = np.random.random() > 0.35  # 65% click rate
    
    # 模拟用户满意度（1-5 星）
    satisfaction = int(np.random.choice([1,2,3,4,5], p=[0.05,0.1,0.2,0.35,0.3]))
    
    ab_results.append({
        'user_id': user_id,
        'group': group,
        'question': result['question'],
        'answer': result['answer'],
        'latency_ms': latency,
        'clicked': click,
        'satisfaction': satisfaction
    })

df_ab = pd.DataFrame(ab_results)

# 统计对比
print('=== A/B 测试结果对比 ===')
for group in ['A', 'B']:
    subset = df_ab[df_ab['group'] == group]
    print(f'\nGroup {group}:')
    print(f'  QPS: {len(subset):.0f} 次/小时')
    print(f'  平均延迟: {subset["latency_ms"].mean():.1f} ms')
    print(f'  点击率: {subset["clicked"].mean():.1%}')
    print(f'  满意度: {subset["satisfaction"].mean():.2f} / 5')

print('\n=== 统计显著性检验（t-test）===')
from scipy import stats
t_stat, p_value = stats.ttest_ind(
    df_ab[df_ab['group']=='A']['satisfaction'],
    df_ab[df_ab['group']=='B']['satisfaction']
)
print(f'  t-statistic: {t_stat:.3f}')
print(f'  p-value: {p_value:.4f}')
print(f'  结论: {"显著 (p < 0.05)" if p_value < 0.05 else "不显著"}')
```

```python
# 失败 case 分析（找出满意度 < 3 星的 case）
failed_cases = df_ab[df_ab['satisfaction'] < 3][['user_id', 'question', 'answer', 'satisfaction']].head(10)

print('=== 失败案例（满意度 < 3）===')
for idx, row in failed_cases.iterrows():
    print(f'\n{idx+1}. Question: {row["question"]}')
    print(f'   Answer: {row["answer"]}')
    print(f'   Satisfaction: {row["satisfaction"]} / 5')
    
# 在 Langfuse 里查这些 trace_id（实际工程里）
print('\n→ 实际工程里：用 trace_id 在 Langfuse 里查完整日志')
```

## 深入思考

1. **A/B 测试的样本量怎么算？**
   - 至少需要 1000 用户，5000 次交互才能有统计显著性
   - 样本量公式：n = (Z^2 * p * (1-p)) / E^2（Z=1.96, E=0.05）
   - p = 预期提升率（比如 5%），E = 允许误差

2. **Langfuse 和其他 APM 的区别？**
   - Langfuse 专精 LLM：trace 包含 embed/retrieve/generate 三个阶段
   - Datadog/Sentry 更通用：也监控后端服务，但对 LLM 的支持不如 Langfuse
   - 中小团队用 Langfuse 开源版够用，大厂可自建

3. **人工评测应该打多少题？**
   - 10-50 题够做 sanity check
   - 100-500 题可以做量化对比
   - 关键是覆盖度（common cases + edge cases）

4. **LLM-as-judge 会一直比人工准吗？**
   - 不一定。LLM-as-judge 有 bias（偏好长答案、偏好礼貌话术）
   - 实证研究表明：LLM-as-judge 和人工相关性约 0.7-0.8
   - 最好的做法：LLM-as-judge 初筛，人工复查异常和关键 case

5. **A/B 测试什么时候应该回滚？**
   - 检测到质量显著下降（p < 0.01，满意度 -0.5 星）
   - 延迟激增（> 2s）
   - 用户投诉增多（电话/邮件 > 5 封/天）
   - 自动化指标全面走低（准确率 < 60%）
